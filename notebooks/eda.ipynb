```python
import json

with open('../data/feedback_raw.json', 'r') as f:
    raw = json.load(f)

# Flatten the nested list
import itertools
flat = list(itertools.chain.from_iterable(raw))

df = pd.DataFrame(flat)
print(df.shape)
df.head()
```

```python
# Check for missing values
print(df.isnull().sum())
```

```python
# Check what the Comment column looks like
print(df['Comment'].head(10))
```

```python
# Keep only the columns we need
df = df[['professor', 'department', 'Quality', 'Difficulty', 'Comment']]

# Convert Quality and Difficulty to numbers
df['Quality'] = pd.to_numeric(df['Quality'], errors='coerce')
df['Difficulty'] = pd.to_numeric(df['Difficulty'], errors='coerce')

# Drop any rows where Comment is empty or whitespace
df = df[df['Comment'].str.strip() != '']

print(df.shape)
df.head()
```

```python
df.to_csv('../data/feedback_clean.csv', index=False)
print("Saved!")
```

```python
import random
random.seed(42)

# Only look at comments where at least one aspect was detected
df_with_aspects = df[df['aspect_count'] > 0].reset_index(drop=True)

# Sample 250 comments to manually label
sample_df = df_with_aspects.sample(n=250, random_state=42).reset_index(drop=True)

# Build one row per (comment, aspect) pair
labeling_rows = []
for idx, row in sample_df.iterrows():
    comment_id = idx
    original_comment = row['Comment']
    aspect_sentiments = row['aspect_sentiments']
    for aspect, (label, score) in aspect_sentiments.items():
        labeling_rows.append({
            'comment_id': comment_id,
            'original_comment': original_comment,
            'aspect': aspect,
            'rule_based_sentiment': label,
            'correct_sentiment': label,   # you'll fix this if wrong
            'aspect_correct': 'yes'       # mark 'no' if this aspect isn't really discussed
        })

labeling_df = pd.DataFrame(labeling_rows)
print(labeling_df.shape)

labeling_df.to_csv('../data/labeling_sheet.csv', index=False)
print("Saved! Open data/labeling_sheet.csv in Excel")
```
